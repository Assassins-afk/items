```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('vgsales.csv')
df.head(10)
```

Задание 2

```python
print("Количество строк:", df.shape[0])
print("Количество столбцов:", df.shape[1])
print("\nНазвания столбцов:")
print(df.columns.tolist())

print("\nТипы данных:")
print(df.dtypes)

print("\nОсновные статистические характеристики:")
print(df.describe())
```

Задание 3

```python
missing = df.isnull().sum()
print("Пропуски по столбцам:")
print(missing[missing > 0])

print("\nСтроки с пропущенным Year:")
print(df[df['Year'].isnull()])
```

Задание 4

```python
df_clean = df.dropna(subset=["Year"])

print("Размер исходного датасета:", df.shape)
print("Размер очищенного датасета:", df_clean.shape)
print("Строк было:", df.shape[0])
print("Строк осталось:", df_clean.shape[0])

print("\nПропуски в df_clean:")
print(df_clean.isnull().sum())
```

Задание 6

```python
genre_count = df_clean.groupby("Genre")["Name"].count().sort_values(ascending=False)
print(genre_count)
```

Задание 7

```python
genre_sales = df_clean.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=False)
print(genre_sales)
```

Задание 8

```python
top10 = df_clean.nlargest(10, "Global_Sales")[["Rank", "Name", "Platform", "Genre", "Global_Sales"]]
print(top10)
```

Задание 9

```python
platform_count = df_clean.groupby("Platform")["Name"].count().sort_values(ascending=False)
print("Топ-10 платформ по количеству игр:")
print(platform_count.head(10))

platform_sales = df_clean.groupby("Platform")["Global_Sales"].sum().sort_values(ascending=False)
print("\nТоп-10 платформ по продажам:")
print(platform_sales.head(10))
```

Задание 10

```python
genre_sales_sorted = genre_sales.sort_values(ascending=True)

plt.figure(figsize=(10, 6))
plt.barh(genre_sales_sorted.index, genre_sales_sorted.values, color='steelblue')
plt.xlabel("Мировые продажи, млн")
plt.title("Суммарные мировые продажи по жанрам")
plt.tight_layout()
plt.show()
```

Задание 13

```python
regions = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales", "Global_Sales"]
corr_matrix = df_clean[regions].corr()
print(corr_matrix)

plt.figure(figsize=(8, 6))
plt.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(label='Корреляция')
plt.xticks(range(len(regions)), regions, rotation=45)
plt.yticks(range(len(regions)), regions)
for i in range(len(regions)):
    for j in range(len(regions)):
        plt.text(j, i, f"{corr_matrix.iloc[i, j]:.2f}",
                 ha='center', va='center', color='black')
plt.title("Корреляционная матрица региональных продаж")
plt.tight_layout()
plt.show()
```
