```python
mean_pop = df['popularity'].mean()
median_pop = df['popularity'].median()
print('Среднее:', mean_pop)
print('Медиана:', median_pop)
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df['popularity'], bins=20, color='skyblue', edgecolor='black')
plt.title('Распределение значений popularity')
plt.xlabel('Popularity')
plt.ylabel('Количество треков')
plt.grid(axis='y', alpha=0.3)
plt.show()
```
